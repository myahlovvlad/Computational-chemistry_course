Теперь зададим другой вопрос: есть ли опубликованная структура комплекса с A2A? Мы читаем метаданные PDB, а не запускаем докинг. Выбор `8RW0` явно зафиксирован, чтобы не смешать его с контрольным примером кофеина `3RFM`.

```python
# for перебирает записи; условие после if выбирает нужную запись по PDB ID.
structure=next(item for item in read_json('structures.json') if item['pdb_id']=='8RW0')
display(structure)
```

У структуры указан метод рентгеновской дифракции, разрешение и первичная публикация. Это другой вид сведений, чем вычисленный дескриптор. Полученная структурная модель помогает исследовать расположение лиганда в изученном комплексе. Разрешение в ангстремах не является концентрацией, константой связывания или точностью клинического прогноза.

Остановитесь и запишите два предложения: «Эта запись поддерживает…» и «Для утверждения о … ещё требуется…». Ниже — один пример, а не единственный допустимый ответ.

```python
claim={
    'question':'Есть ли опубликованное структурное свидетельство комплекса истрадефиллина с A2A?',
    'supported':'В PDB опубликована структура 8RW0 с указанными методом и первичной статьёй.',
    'not_established':'Из этих метаданных нельзя определить проникновение через ГЭБ или клеточную экспозицию.',
    'next_evidence':'Для вопроса о проникновении нужны подходящие данные экспозиции/проницаемости с условиями и контролями.',
    'sources':[card['source_url'],'https://www.rcsb.org/structure/8RW0'],
    'uncertainty':'В этом уроке не исследованы координаты, качество плотности и функциональные измерения.'
}
display(claim)
write_json(OUT/'claim_and_evidence.json',claim)
```

Мы не обесценили структуру, назвав ограничение. Мы уточнили её место в исследовании. Позднее вернёмся к координатам и функциональным тестам; пока научились не перепрыгивать через недостающий переход.

Междисциплинарная карта начинается с одного объяснённого ребра. Оно связывает два понятия и указывает, на каком основании связь проведена. Полезное ребро содержит смысл; стрелка без объяснения только украшает схему.

```python
edge={
    'from':'структурная биология','to':'гипотеза о связывании с A2A',
    'explanation':'Структура комплекса показывает расположение лиганда в изученной конструкции белка.',
    'source':'https://www.rcsb.org/structure/8RW0',
    'limit':'Не измеряет тканевую экспозицию и не переносит активность на другие молекулы.',
    'next_check':'Позже сопоставить структуру с качеством модели и сопоставимыми функциональными тестами.'
}
write_json(OUT/'concept_edge.json',edge)
print('Сохранены пример интерпретации и ребро карты:',OUT)
```
